# 6. Análise final dos resultados

## 6.1 Comparação das características extraídas
A comparação entre as quatro configurações de características permitiu avaliar a contribuição dos atributos de cor, textura, forma e HOG para o desempenho dos classificadores. Os resultados demonstraram que a inclusão de uma quantidade maior de características não proporcionou necessariamente melhorias para todos os modelos.

Para o classificador KNN, a configuração B, composta por características de cor e textura, apresentou o melhor resultado, com F1-score macro de 99,09%. A inclusão das características de forma e HOG resultou em uma redução do desempenho, indicando que a utilização de mais atributos não foi necessariamente vantajosa para esse classificador. Uma possível explicação está relacionada à influência das características adicionais no cálculo das distâncias entre as amostras.

No caso do SVM, observou-se uma evolução do desempenho à medida que novas características foram incorporadas. O F1-score macro passou de 92,78% na configuração A para 99,23% na configuração D, que reuniu cor, textura, forma e HOG. Esse comportamento sugere que o classificador conseguiu aproveitar informações complementares presentes nos diferentes grupos de características.

O Random Forest apresentou desempenho elevado em todas as configurações, alcançando F1-score macro de 99,78% tanto na configuração C quanto na D. Dessa forma, a inclusão das características HOG não proporcionou uma melhoria expressiva em relação à combinação de cor, textura e forma. Considerando que a configuração C utiliza apenas 52 características, enquanto a D utiliza 376, a configuração C apresenta uma representação mais compacta, mantendo desempenho equivalente nessa avaliação.

De maneira geral, os resultados evidenciam que a contribuição das características depende do classificador utilizado. Enquanto o KNN apresentou melhor desempenho com uma representação mais simples, o SVM se beneficiou da combinação de todos os grupos de características. Já o Random Forest alcançou resultados praticamente equivalentes sem a necessidade de incorporar o HOG. Assim, a escolha da representação mais adequada deve considerar não apenas o desempenho preditivo, mas também a quantidade de atributos e o custo computacional envolvido.


## 6.2 Comparação dos classificadores
A avaliação final dos classificadores foi realizada utilizando o conjunto de teste, composto por 12.528 imagens, mantendo os dados separados durante as etapas de treinamento e otimização dos hiperparâmetros. Para cada algoritmo, foi utilizada a configuração de características que apresentou melhor desempenho nos experimentos anteriores.

O KNN foi avaliado com as características de cor e textura, totalizando 39 atributos. O SVM utilizou a configuração composta por cor, textura, forma e HOG, com 376 atributos. Já o Random Forest foi treinado com a configuração de cor, textura e forma, totalizando 52 atributos.

Os resultados demonstraram que o Random Forest apresentou o melhor desempenho geral, alcançando acurácia de 98,67% e F1-score macro de 98,61%. O KNN também apresentou resultados elevados, com acurácia de 98,44% e F1-score macro de 98,41%, ficando próximo do desempenho do Random Forest.

O SVM obteve acurácia de 97,25% e F1-score macro de 97,08%. Embora tenha apresentado resultados inferiores aos outros dois classificadores, seu desempenho ainda demonstra boa capacidade de reconhecimento das classes presentes no conjunto de teste.

A proximidade entre os resultados do Random Forest e do KNN indica que ambos conseguiram representar adequadamente as características visuais das frutas. Entretanto, considerando o conjunto das métricas avaliadas, o Random Forest apresentou o melhor resultado, além de utilizar uma quantidade menor de atributos que a configuração adotada pelo SVM.

Dessa forma, o Random Forest foi considerado o classificador de melhor desempenho na avaliação final, seguido pelo KNN e pelo SVM.


## 6.3 Análise das classes confundidas
A análise das matrizes de confusão permitiu identificar quais classes apresentaram maior dificuldade de classificação e quais pares de frutas foram frequentemente confundidos pelos modelos.

Entre os erros observados, algumas confusões apareceram em mais de um classificador. Um exemplo foi a classificação de Pepper Yellow 1 como Pepper Orange 2, identificada nos três modelos: KNN, SVM e Random Forest. Esse comportamento pode estar relacionado à semelhança visual entre as classes, especialmente em relação à coloração e às características da superfície.

Também foram observadas confusões entre variedades de cerejas, como Cherry 2 e Cherry 4, principalmente nos resultados do KNN e do Random Forest. Outro caso recorrente envolveu Apple Pink Lady 1 e Apple Red 2, observado principalmente nos resultados do KNN e do Random Forest, indicando dificuldades na distinção entre algumas variedades de maçãs.

No caso do milho, o KNN e o Random Forest classificaram imagens da classe Corn 1 como Physalis with Husk 1. Já o SVM apresentou confusão entre Corn 1 e Onion White 1. Esses resultados mostram que os classificadores podem apresentar padrões distintos de erro, mesmo quando trabalham com o mesmo conjunto de imagens.

Também foram identificadas confusões entre variedades de batata. O KNN e o Random Forest classificaram imagens de Potato Sweet 1 como Potato Red 1, enquanto o SVM apresentou confusão com Potato White 1.

De modo geral, os erros observados indicam que algumas classes possuem características visuais semelhantes, o que pode dificultar sua separação a partir dos atributos extraídos. Entretanto, essas possíveis explicações devem ser consideradas hipóteses, pois a análise dos resultados, isoladamente, não comprova quais características específicas causaram cada confusão.

Embora o Random Forest tenha apresentado o melhor desempenho geral na avaliação final, os resultados demonstram que nenhum classificador eliminou completamente as dificuldades de distinção entre determinadas classes.


## 6.4 Contribuição do pré-processamento e da segmentação
As etapas de pré-processamento e segmentação foram importantes para organizar as imagens e preparar os dados para a extração de características. Como o conjunto Fruits-360 já disponibilizava imagens com resolução de 100 × 100 pixels, não foi necessário realizar redimensionamento.

Também optou-se por não aplicar filtros de remoção de ruído, considerando que esse procedimento poderia eliminar detalhes importantes da superfície das frutas, como textura e pequenas variações visuais. Foi implementada a possibilidade de utilização da técnica CLAHE no canal de luminosidade do espaço de cores Lab, com o objetivo de melhorar o contraste local das imagens. Entretanto, essa configuração era opcional e não deve ser considerada aplicada a todas as imagens do experimento.

Na etapa de segmentação, foram utilizadas técnicas de limiarização de Otsu, considerando informações em escala de cinza e no canal de saturação do espaço HSV. As máscaras obtidas foram combinadas e submetidas a operações morfológicas de fechamento, além da identificação e preenchimento do maior contorno. O objetivo foi separar a região correspondente à fruta do fundo da imagem, favorecendo a extração de características relacionadas ao objeto.

A segmentação também contribuiu para orientar a extração de atributos de cor, textura e forma, permitindo concentrar a análise na região identificada como objeto. Durante esse processo, foram identificados casos de máscaras vazias, tratados por meio de uma variável de controle, que não foi utilizada como característica de entrada dos classificadores.

Apesar da importância dessas etapas na construção do pipeline, não foi realizado um experimento quantitativo específico comparando o desempenho dos classificadores com e sem segmentação ou com e sem pré-processamento. Portanto, não é possível afirmar, com base nos resultados obtidos, que essas técnicas provocaram um aumento direto nas métricas de classificação.

Assim, sua contribuição deve ser compreendida principalmente como parte da preparação e organização das imagens para a extração de características, enquanto a mensuração de seu impacto individual permanece como uma possibilidade para experimentos futuros.

## 6.5 Discussão geral dos resultados

Os experimentos realizados demonstraram que a classificação de imagens de frutas utilizando características extraídas manualmente e algoritmos tradicionais de aprendizado de máquina apresentou resultados bastante satisfatórios. Entre os classificadores avaliados, o Random Forest obteve o melhor desempenho na avaliação final, alcançando 98,67% de acurácia e 98,61% de F1-score macro. O KNN apresentou resultados próximos, com 98,44% de acurácia e 98,41% de F1-score macro, enquanto o SVM obteve 97,25% de acurácia e 97,08% de F1-score macro.

A comparação entre os conjuntos de características evidenciou que a combinação de atributos influencia o desempenho de maneira diferente para cada classificador. O KNN apresentou melhores resultados com a combinação de cor e textura, enquanto o SVM obteve seu melhor desempenho com a inclusão dos atributos HOG. Para o Random Forest, a configuração composta por cor, textura e forma apresentou desempenho praticamente equivalente à configuração que também incluía HOG, indicando que o acréscimo desses atributos não proporcionou ganho relevante nos resultados observados.

A análise dos erros também revelou que determinadas variedades de frutas apresentaram maior dificuldade de distinção, especialmente aquelas com características visuais semelhantes. Foram identificadas confusões recorrentes entre variedades de pimentões, cerejas, maçãs e batatas. Esses resultados mostram que, mesmo com métricas gerais elevadas, ainda existem situações em que os atributos extraídos não são suficientes para distinguir completamente determinadas classes.

Em relação ao pré-processamento e à segmentação, essas etapas foram incorporadas ao pipeline para preparar as imagens e orientar a extração de características. Entretanto, como não foi realizada uma comparação experimental isolando a contribuição dessas técnicas, não é possível quantificar seu impacto individual sobre o desempenho dos classificadores.
